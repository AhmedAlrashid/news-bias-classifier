### RoBERTa Pretrained Model Demonstration

```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
```

```python
model = AutoModelForSequenceClassification.from_pretrained('avanishd/roberta-pretrained-ground-news')
tokenizer = AutoTokenizer.from_pretrained('avanishd/roberta-pretrained-ground-news')
```

```python
# Link to samples
# 1. https://www.tampafp.com/federal-judge-in-massachusetts-blocks-rfk-jr-s-plan-to-slash-recommended-childhood-vaccines/#google_vignette
# 2. https://www.militarytimes.com/news/your-military/2026/03/16/number-of-us-troops-wounded-in-war-against-iran-rises-to-about-200/
# 3. https://www.nbcnews.com/politics/trump-administration/us-wants-delay-china-trip-xi-jinping-meeting-month-due-iran-war-rcna263792

samples = [
    """ A federal judge in Boston stepped in on Monday to stop the U.S. Health Department from cutting back the list of vaccines recommended for children across the country. According to a report by the Associated Press, Judge Brian E. Murphy issued a temporary stay against Health Secretary Robert F. Kennedy Jr., ruling that the administration likely broke federal rules when it tried to overhaul a major vaccine advisory board and strip several shots from the national schedule. """,
    """ WASHINGTON, March 16 (Reuters) — The number of U.S. troops wounded in the war against Iran has risen to about 200, the U.S. military said on Monday, as the conflict entered its third week. The U.S. military’s Central Command said the vast majority of those wounded had suffered minor injuries and 180 troops had already returned to duty. Ten of the injuries are serious, it said. """, 
    """ President Donald Trump said on Monday that the U.S. requested that his trip to China be delayed by “a month or so,” citing the war in Iran as a reason. Trump told reporters in the Oval Office that he didn’t know whether he still planned to travel to Beijing to meet with Chinese President Xi Jinping at the end of the month as previously scheduled."""
]
```

```python
inputs = tokenizer(samples, return_tensors='pt', truncation=True, padding=True, max_length=512)
```

```python
int_to_label = {0: "Far Left", 1: "Left", 2:"Lean Left", 3:"Center", 4:"Lean Right", 5:"Right", 6:"Far Right"}
```

```python
with torch.no_grad():
    logits = model(**inputs).logits
    pred_ids = torch.argmax(logits, dim=-1).tolist()

labels = [int_to_label[i] for i in pred_ids]
print(labels)
```
